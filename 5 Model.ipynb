
###  **Model 1**

**Transformations / Nonlinearity:**

* `age_squared`: Captures the nonlinear (possibly diminishing) effect of age on wage.
* `hours_squared`: Models diminishing returns to hours worked.

**Interaction Effects:**

* `educ_IQ`: Tests whether higher IQ amplifies the return to education.
* `married_exper`: Captures how marriage may enhance the value of work experience.

**Forced Inclusion:**

* `educ` is explicitly included.

---

###  **Model 2**

**Transformations / Nonlinearity:**

* `age_4`, `educ_4`: 4th-order polynomials capture more complex nonlinearities in age and education.
* `log_hours`, `log_tenure`: Applied to stabilize variance and compress skewed distributions.

**Interaction Effects:**

* `IQ_educ`: IQ may reinforce the marginal productivity of education.
* `married_exper`: Combines marital status and experience to reflect enhanced job stability or return.

**Forced Inclusion:**

* `educ` is included directly and in the interaction term; `exper` is involved in `married_exper`.

---

###  **Model 3**

**Transformations / Nonlinearity:**

* `log_educ`, `log_hours`, `log_tenure`: Implements a log-log specification to estimate elasticities and mitigate skewness.

**Interaction Effects:**

* `IQ_educ`: Reflects multiplicative gains in productivity when both IQ and education are high.
* `married_exper`: Evaluates whether being married alters the return to experience.

**Forced Inclusion:**

* `educ` is included through `log_educ` and in `IQ_educ`.

---

###  **Model 4**

**Transformations / Nonlinearity:**

* `log_educ`, `log_tenure`: Log transformations smooth nonlinear effects and help with heteroskedasticity.
* `hours_spline_35`, `hours_spline_45`: Piecewise linear (spline) terms allow for changes in the marginal effect of hours at critical thresholds (e.g., 35 and 45 hours per week).

**Interaction Effects:**

* `IQ_educ`: Captures compound productivity from cognitive ability and formal education.
* `married_exper`: Reflects how marriage may shift the experience–wage relationship.

**Forced Inclusion:**

* `educ` is included as `log_educ` and in interaction.

---

###  **Model 5**

**Transformations / Nonlinearity:**

* `log_educ`, `log_hours`, `log_tenure`: Stabilize distributions and estimate wage elasticities.
* `spline_hours_30`, `spline_hours_40`, `spline_hours_50`: Introduce flexible nonlinear shifts in the wage–hours relationship, capturing varying returns to working longer hours.

**Interaction Effects:**

* `IQ_educ`: Allows IQ to amplify educational returns.
* `married_exper`: Marriage could buffer or intensify how experience translates to wages.

**Forced Inclusion:**

* `educ` is present both directly (via logs) and in interaction; `exper` is used in `married_exper`.
